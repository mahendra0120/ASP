```python
import base64
import pandas as pd
import numpy as np
from pathlib import Path
from natsort import natsorted
from IPython.display import display, Image, HTML


def _single_img_html(path) -> str:
    p = Path(path)
    if not p.is_file():
        return f"<span>{p}</span>"
    b64 = base64.b64encode(p.read_bytes()).decode("ascii")
    ext = p.suffix.lower()
    mime = "image/png" if ext == ".png" else "image/jpeg" if ext in (".jpg", ".jpeg") else "image/png"
    return f'<img src="data:{mime};base64,{b64}" style="max-width:640px;height:auto;display:block"/>'

def path_to_img_html(path_or_paths) -> str:
    """One path or a list/tuple of paths → stacked HTML images for Styler."""
    if path_or_paths is None or (isinstance(path_or_paths, float) and pd.isna(path_or_paths)):
        return ""
    if isinstance(path_or_paths, (str, Path)):
        return _single_img_html(path_or_paths)
    try:
        paths = list(path_or_paths)
    except TypeError:
        return _single_img_html(path_or_paths)
    if not paths:
        return ""
    imgs = "".join(_single_img_html(p) for p in paths)
    return f'<div style="display:flex;flex-direction:column;gap:8px;align-items:flex-start;">{imgs}</div>'
```

```python
import textwrap
import re

def format_to_html(items):
    def style_line(line):
        styled = re.sub(r'\b([A-Z]{2,})\b', r'<b>\1</b>', line)
        if re.search(r'^[A-Z\s,/:]{5,}', line):
            return f"<div style='text-align:center;'>{styled}</div>"
        return f"<div style='text-align:left;'>{styled}</div>"

    html_parts = []
    for item in items:
        lines = [l.strip() for l in textwrap.dedent(item).strip().splitlines() if l.strip()]
        if not lines:
            continue
        styled_lines = "".join(style_line(l) for l in lines)
        html_parts.append(styled_lines)
    return "<br>".join(html_parts)

for i in range(0, 33):
    globals()[f"brief{i}"] = format_to_html(globals()[f"brief{i}"])
    globals()[f"exam{i}"]  = format_to_html(globals()[f"exam{i}"])

rows = []
for f, b, e in zip(list(globals()[f"flist{i}"] for i in range(0, 33)), list(globals()[f"brief{i}"] for i in range(0, 33)), list(globals()[f"exam{i}"] for i in range(0, 33))):
    rows.append({
        "Autopsy Photos": f,
        "Brief Summary":  b,
        "Examination":    e,
    })

aut_dat = pd.concat([
    pd.DataFrame({
        "Autopsy Number": [f"Autopsy Number: {i}" for i in range(1, 34)]
    }),
    pd.DataFrame(rows),
    pd.DataFrame({
        "Cause of Death":                  [globals()[f"cod{i}"]  for i in range(0, 33)],
        "Manner of Death":                 [globals()[f"mod{i}"]  for i in range(0, 33)],
        "Murder Weapon":                   [globals()[f"mw{i}"]   for i in range(0, 33)],
        "Reported Circumstances of Death": [globals()[f"rcod{i}"] for i in range(0, 33)],
        "General External Examination":    [globals()[f"gee{i}"]  for i in range(0, 33)],
        "Evidence of Trauma":              [globals()[f"eot{i}"]  for i in range(0, 33)],
    }),
], axis=1)

col_widths = {
    "Autopsy Number":                  "120px",
    "Autopsy Photos":                  "660px",
    "Brief Summary":                   "220px",
    "Examination":                     "420px",
    "Cause of Death":                  "160px",
    "Manner of Death":                 "120px",
    "Murder Weapon":                   "120px",
    "Reported Circumstances of Death": "220px",
    "General External Examination":    "240px",
    "Evidence of Trauma":              "160px",
}

styles = [
    {
        "selector": f"th.col_heading.col{i}, td.col{i}",
        "props": [
            ("width",          w),
            ("min-width",      w),
            ("max-width",      w),
            ("word-wrap",      "break-word"),
            ("white-space",    "pre-wrap"),
            ("vertical-align", "middle"),
            ("text-align",     "center"),
            ("padding",        "6px 8px"),
        ],
    }
    for i, (col, w) in enumerate(col_widths.items())
]

html = (
    aut_dat.style
    .format({
        "Autopsy Photos": path_to_img_html,
        "Brief Summary":  lambda x: x,
        "Examination":    lambda x: x,
    })
    .set_table_styles(styles)
    .hide(axis="index")
    .to_html()
)

display(HTML(f'<div style="overflow-x:auto;">{html}</div>'))
```
